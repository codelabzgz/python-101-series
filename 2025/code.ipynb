## 0. Sintaxis

### Estructura y Bloques

A diferencia de C/Java/JS, Python no usa llaves `{}` para definir bloques de código. Usa **indentación estricta**.

- **Indentación:** 4 espacios es el estándar ([PEP 8](https://peps.python.org/pep-0008/)). Mezclar tabs y espacios lanzará un `IndentationError`.
- **Comentarios:** `#` para una línea. `""" Docstrings """` para documentación multilinea.
- **Variables:** No se declaran, se asignan (y sin tipo explícito!). La convención de nombrado es `snake_case`.

### Algunos operadores

- División flotante: `5 / 2` -> `2.5`
- División entera: `5 // 2` -> `2`
- Potencia: `2 ** 3` -> `8` (No usar `^` que es XOR bit a bit)
- Lógica booleana: `and`, `or`, `not` (en lugar de `&&`, `||`, `!`)


````python
# Asignación múltiple (muy común en Python)
a, b = 5, 2

print(f"División real: {a / b}")  # 2.5
print(f"División entera: {a // b}")  # 2
print(f"Potencia: {a ** b}")  # 25

# Intercambio de variables sin variable temporal
a, b = b, a
print(f"Swap -> a: {a}, b: {b}")
````

## 1. Modelo de datos

En Python, las variables no son "cajas" que contienen datos. Son etiquetas (referencias) que apuntan a objetos en memoria.

### Tipado

- Tipado dinámico: El tipo se asigna en _tiempo de ejecución_
- Pero **tipado fuerte**: Las conversiones de tipo están _muy_ limitadas


````python
# Type Hints (Python 3.5+): Son informativos para el IDE, no fuerzan el tipo en runtime.
x: int = 10
y = x  # y apunta al mismo objeto que x

print(f"ID memoria x: {id(x)}")
print(f"ID memoria y: {id(y)}")

# Los enteros son inmutables. Al 'cambiar' x, creamos un nuevo objeto.
x = x + 1
print(f"Nuevo ID x: {id(x)}")
print(f"El ID de y se mantiene: {id(y)}")

# Tipado Fuerte
try:
    print("El número es " + 10)
except TypeError as e:
    print(f"Error: {e}")
    print("Correcto: 'El número es ' + str(10)")
````

## 2. Strings

Python tiene de forma nativa una de las manipulaciones de texto más potentes.

### Propiedades

- **Slicing:** `str[start : end : step]`
- **Inmutabilidad:** Los strings **no** se pueden modificar in-place


````python
text = "Python Workshop"

# Slicing (El límite superior siempre es EXCLUSIVO)
print(text[0:6])  # 'Python'
print(text[:6])  # 'Python' (asume 0 al inicio)
print(text[-1])  # 'p' (último carácter)
print(text[::-1])  # 'pohskroW nohtyP' (Invertir string idiomático)

# Métodos útiles
raw_data = "  admin,1234,active  "
clean = raw_data.strip().split(",")  # Encadenado de métodos
print(f"Datos: {clean}")

# F-Strings (Python 3.6+): La forma moderna de formatear
user = "Admin"
logs = 150
print(f"User: {user} | Logs: {logs} | Hex: {logs:04x}")
````

## 3. Estructuras de control de flujo

Antes de manipular datos, necesitamos lógica. En Python, la indentación es sintaxis, no estilo.

### Condicionales (if, elif, else)

- No se usan paréntesis `()` alrededor de la condición.
- **Truthiness:** Concepto vital en Python. Los contenedores vacíos (`[]`, `{}`, `""`), el número `0` y `None` se evalúan como `False`. Todo lo demás es `True`. No escribas `if len(lista) > 0`


````python
status = 200
data = []

# Sintaxis estándar
if status == 200:
    print("Success")
elif status == 500:
    print("Server Error")
else:
    print("Unknown status")

# Evaluación 'Pythonic' de vacíos
# NO hacer: if len(data) == 0:
if not data:
    print("La lista está vacía (False implícito)")
````

### Bucles básicos

El bucle `for` en Python actúa como un bucle `foreach` en otros lenguajes, iterando directamente sobre los elementos de cualquier iterable (lista, tupla, string, etc).


````python
valores = [10, 50, 100]

for v in valores:
    if v > 40:
        print(f"Valor alto: {v}")
````

El bucle `while` es idéntico al de otros lenguajes, evaluando la condición antes de cada iteración.


````python
count = 5
while count > 0:
    print(f"Countdown: {count}")
    count -= 1
    # Nota: Python NO tiene operadores de incremento/decremento (count-- o count++)
    # Debes usar count -= 1
````

Sin embargo, Python **no** tiene una estructura nativa `do-while`. La forma idiomática (_Pythonic_) de garantizar que el código se ejecute al menos una vez es usar un bucle infinito while True con una condición de salida explícita (break) al final del bloque.


````python
# Patrón para simular do-while (ej: Menú de usuario)
while True:
    comando = input("Escribe 'salir' para terminar: ")
    print(f"Procesando: {comando}")

    if comando == "salir":
        break
````

Python permite un bloque else asociado a un `while` (o `for`), que se ejecuta solo si el bucle termina "naturalmente" (cuando la condición se vuelve falsa). Este bloque **no** se ejecuta si el bucle fue interrumpido por un break.

Esto es muy útil para búsquedas, evitando la típica variable bandera `bool found = false;` de C++.


````python
# Búsqueda sin "flags" booleanas
target = 5
items = [1, 3, 7, 9]
i = 0

while i < len(items):
    if items[i] == target:
        print("Encontrado!")
        break
    i += 1
else:
    # Solo entra aquí si NO se ejecutó el break
    print("El elemento no está en la lista")
````

## 4. Estructuras de datos fundamentales

Python es un lenguaje multiparadigma, tomando buenas ideas de otros lenguajes como los funcionales o los orientados a objetos.

### Listas y Tuplas

- Lista `[]`: Mutable, es en esencia un array dinámico
- Tupla `()`: Inmutable, más ligera en memoria y utilizada para datos fijos


````python
# --- LISTAS ---
stack = [1, 2, 3]
stack.append(4)  # Push O(1)
last = stack.pop()  # Pop O(1)

# List Slicing con asignación
nums = [10, 20, 30, 40]
nums[1:3] = [99, 99]  # Reemplazo in-place
print(f"Lista mutada: {nums}")

# --- TUPLAS ---
config = ("localhost", 8080)
# config[0] = "127.0.0.1"  # TypeError: 'tuple' object does not support item assignment

# Unpacking (Desempaquetado)
host, port = config
print(f"Host: {host}, Port: {port}")

# Desempaquetado extendido (*)
first, *middle, last = [1, 2, 3, 4, 5, 6]
print(f"Middle: {middle}")  # [2, 3, 4, 5]
````

````python
colors = ["red", "green", "blue"]

# 1. Iteración directa (foreach implícito)
for c in colors:
    pass

# 2. Si necesitas el índice: enumerate()
for i, c in enumerate(colors):
    print(f"Index {i}: {c}")

# 3. Iterar dos listas en paralelo: zip()
codes = [255, 128, 0]
for color, code in zip(colors, codes):
    print(f"{color} -> {code}")
````

### Otras estructuras: Diccionarios y Sets

Acceso promedio O(1). Las claves deben ser objetos hashables (inmutables).


````python
# --- DICCIONARIOS ---
user = {"id": 1, "name": "Dev", "role": "admin"}

# Acceso seguro .get() evita try-catch por KeyError
role = user.get("permissions", "read-only")
print(f"Permisos: {role}")

# --- SETS ---
# Útiles para eliminar duplicados y lógica de conjuntos
ids_a = {1, 2, 3, 4}
ids_b = {3, 4, 5, 6}

print(f"Intersección: {ids_a & ids_b}")  # {3, 4}
print(f"Diferencia: {ids_a - ids_b}")  # {1, 2}
````

## 5. Comprensión de listas

La característica funcional más usada. Permite crear listas/diccionarios nuevos a partir de iterables en una sola línea (más o menos) legible.

`[expresion for item in iterable if condicion]`


````python
raw_nums = [1, 2, 3, 4, 5, 6]

# Enfoque clásico (Imperativo)
squares = []
for n in raw_nums:
    if n % 2 == 0:
        squares.append(n**2)

# List Comprehension (Declarativo)
squares_comp = [n**2 for n in raw_nums if n % 2 == 0]

print(squares_comp)  # [4, 16, 36]

# Dictionary Comprehension
# Crear mapa {numero: cuadrado}
sq_dict = {n: n**2 for n in raw_nums}
print(sq_dict)
````

## Funciones

Las funciones son objetos de primera clase


````python
# Argumentos con valores por defecto y *args/**kwargs
def logger(msg, level="INFO", *args, **kwargs):
    print(f"[{level}] {msg} | Extras: {args} {kwargs}")


logger("System start", "WARN", 101, user="root")


# ⚠️ EL PELIGRO DE LOS MUTABLES ⚠️
# NUNCA uses una lista o dict vacío como valor por defecto.
# Se evalúa una sola vez al definir la función, no al llamarla.


def add_bad(item, lista=[]):
    lista.append(item)
    return lista


print(add_bad("A"))  # ['A']
print(add_bad("B"))  # ['A', 'B'] ¡Mantiene estado!


# Patrón correcto:
def add_good(item, lista=None):
    if lista is None:
        lista = []
    lista.append(item)
    return lista
````

## 7. Gestión de recursos (Context Managers)

La sentencia `with` asegura que los recursos (archivos, sockets, conexiones DB) se cierren o liberen, ocurra o no una excepción.


````python
# Manera insegura
f = open("test.txt", "w")
f.write("Hola")
# Si falla algo aquí, f.close() nunca se ejecuta
f.close()

# Manera Pythonic (Context Manager)
with open("test.txt", "w") as f:
    f.write("Data segura")
    # El archivo se cierra automáticamente al salir del bloque

print(f"¿Cerrado?: {f.closed}")
````

## Bonus


### Gestión de errores (EAFP)

Python sigue el principio **EAFP** (Easier to Ask for Forgiveness than Permission). En lugar de comprobar si un archivo existe antes de abrirlo, intenta abrirlo y captura el error.

- `try`: Bloque de código propenso a errores
- `except`: Captura la excepción (puedes especificar el tipo)
- `finally`: Se ejecuta siempre (limpieza)


````python
data = {"a": 1}

try:
    # Intentamos acceder a una clave que no existe
    print(data["b"])
except KeyError as e:
    print(f"Error capturado: Clave {e} no encontrada.")
except Exception as e:
    # Catch-all (equivalente a catch(...) en C++)
    print(f"Error genérico: {e}")
finally:
    print("Operación finalizada.")
````

### Clases y objetos

Python es orientado a objetos. Todo (enteros, funciones, tipos) es un objeto.

- **`__init__`**: Es el constructor
- **`self`**: Es el equivalente a `this`, pero **debe ser explícito** como primer argumento de los métodos de instancia
- **`__str__`**: Equivalente a sobrecargar el operador `<<` de C++ para imprimir o `toString()`


````python
class Developer:
    # Atributo de clase (static)\n",
    company = "Tech Corp"

    def __init__(self, name: str, lang: str):
        # Atributos de instancia\n",
        self.name = name
        self.lang = lang

    def code(self):
        return f"{self.name} está escribiendo {self.lang}"

    def __str__(self):
        return f"[Dev: {self.name}]"


dev = Developer("Alice", "Python")
print(dev.code())
print(dev)  # Usa __str__"
````

### Decoradores

Los decoradores son _azúcar sintáctico_ (`@`) para aplicar **Funciones de órden superior**. Permiten extender el comportamiento de una función sin modificar su código fuente (similar al patrón Wrapper/Proxy pero mediante una aproximación funcional).
Técnicamente, un decorador es una función que recibe una función y devuelve otra función (normalmente un _closure_ que envuelve a la original)

```python
    @mi_decorador
    def funcion():
        pass
    # Es equivalente a
    funcion = mi_decorador(funcion)
```


````python
import time


# Definición del decorador
def timer_decorator(func):
    # Definimos el wrapper que aceptará cualquier argumento (*args, **kwargs)
    def wrapper(*args, **kwargs):
        start = time.time()
        print(f"[START] Ejecutando {func.__name__}...")

        # Ejecutamos la función original
        result = func(*args, **kwargs)
        end = time.time()
        print(f"[END] Tiempo total: {end - start:.4f}s")
        return result

    return wrapper


# Uso del decorador\n",
@timer_decorator
def process_data(n):
    time.sleep(1)  # Simulamos trabajo pesado
    return n * n


# Llamada
res = process_data(10)
print(f"Resultado: {res}")
````
