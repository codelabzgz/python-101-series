# 1. Variables y tipos

Las variables nos permiten guardar información para utilizarla posteriormente.

En Python no es necesario indicar explícitamente el tipo de una variable. El tipo se determina a partir del valor que le asignamos.

Algunos de los tipos básicos que utilizaremos son:

* `int` — números enteros
* `float` — números decimales
* `str` — cadenas de texto
* `bool` — valores booleanos: `True` o `False`

Podemos comprobar el tipo de un valor utilizando la función `type()`.


```python
nombre = "Pepe Potamo"
edad = 20
altura = 1.72
estudiante = True

print(nombre)
print(edad)
print(altura)
print(estudiante)
```

```python
print(type(nombre))
print(type(edad))
print(type(altura))
print(type(estudiante))
```

## Operaciones con números

Python permite realizar las siguientes operaciones matemáticas:

* `+` suma
* `-` resta
* `*` multiplicación
* `/` división
* `//` división entera
* `%` resto de la división
* `**` potencia


```python
a = 10
b = 3

print(a + b)   # Suma
print(a - b)   # Resta
print(a * b)   # Multiplicación
print(a / b)   # División
print(a // b)  # División entera
print(a % b)   # Resto
print(a ** b)  # Potencia
```

```python
# Tambien podemos guardar los valores en variables

a = 10
b = 3

suma = a + b
resto = a % b

print("Suma:", suma)
print("Resto:", resto)
```

# 2.  Condicionales

Los condicionales permiten que un programa tome decisiones. Dependiendo de si una condición es verdadera o falsa, podemos ejecutar un bloque de código u otro.

Para construir condiciones utilizaremos operadores de comparación como:

* `==` igual
* `!=` distinto
* `<` menor que
* `>` mayor que
* `<=` menor o igual
* `>=` mayor o igual

Las condiciones producen un valor booleano: `True` o `False`.


```python
edad = 20

print(edad >= 18)
print(edad == 20)
print(edad < 18)
```

### Bloques condicionales

`if` permite ejecutar un bloque de código solamente cuando una condición se cumple.



`else` permite ejecutar un bloque de código solamente cuando una condición no se ha cumplido.

```python
edad = int(input("Cuántos años tienes?\n"))

if edad >= 18:
    print("Puedes entrar")
else:
    print("No puedes entrar")
```

`elif` permite ejecutar un bloque de código sólamente cuando una condición no se ha cumplido, pero hay otra que sí.

Las condiciones en un bloque `if`, `elif`, `else`, se van comprobando de arriba a abajo.

```python
nota = int(input("Qué nota has sacado?\n"))

if nota >= 9:
    print("Sobresaliente")
elif nota >= 5:
    print("Aprobado")
else:
    print("Suspenso")
```

### `and`, `or` y `not`

Podemos combinar varias condiciones utilizando operadores booleanos:

* `and` → se deben cumplir ambas condiciones.
* `or` → se debe cumplir al menos una.
* `not` → invierte el resultado de una condición.


```python
edad = int(input("Cuántos años tienes?\n"))
tiene_entrada = True

if edad >= 18 and tiene_entrada:
    print("Puedes entrar")
```

```python
es_alumno = True
es_profesor = False

if es_alumno or es_profesor:
    print("Puedes acceder")
```

# 3.  `while`

El bucle `while` permite repetir un bloque de código **mientras una condición sea verdadera**.

En cada iteración se vuelve a comprobar la condición.

```python
numero = 1

while numero <= 5:
    print(numero)
    numero += 1 # Qué pasa si quitamos esta línea?
```

- `while True` crea un bucle que, en principio, no termina.

- `break` permite salir del bucle cuando nosotros decidimos.

```python
# Bucle "infinito"
while True:
    texto = input("Escribe algo: ")

    if texto == "salir":
        break

    print(f"Has escrito: {texto}")
```

# 5. Listas y tuplas

Las **listas** permiten almacenar varios valores en una misma variable. Pueden contener valores de cualquier tipo y podemos modificarlas después de crearlas.

Los elementos de una lista tienen una posición, empezando por `0`.

Las **tuplas** son parecidas a las listas, pero sus elementos no se pueden modificar después de crearlas.


```python
notas = [4, 7, 8, 3, 9]

print(notas) # imprimir toda la lista
print(notas[0]) # acceder a elementos sueltos
print(notas[2])
```

```python
# Esto compila?
print(notas[-1])
```

```python
# Modificar listas

notas[0] = 5 # Cambiar un elemento
notas.append(10) # Añadir un elemento al final

print(notas)
```

### Funciones habituales de listas

```python
print(len(notas))
print(sum(notas))
print(max(notas))
print(min(notas))
```

### Slicing

Podemos crear una lista a partir de los elementos de otra

```python
print(notas[1:4])
```

### Recorrer una lista

```python
for nota in notas:
    print(nota)
```

Usando `enumerate()` podemos ver el indice asociado a cada elemento

```python
for i, nota in enumerate(notas):
    print(i, nota)
```

# 6. Diccionarios y sets

Los **diccionarios** permiten almacenar información asociando una **clave** con un **valor**.

A diferencia de las listas, no accedemos a los elementos mediante una posición, sino mediante su clave.

Se escriben utilizando llaves `{}` y cada elemento tiene la forma `clave: valor`.


```python
alumno = {
    "nombre": "Ana",
    "edad": 20,
    "nota": 8.5
}

print(alumno)
print(alumno["nombre"])
print(alumno["nota"])
```

```python
# Los campos son modificables
alumno["nota"] = 9
alumno["curso"] = 1

print(alumno)
```

>Para iterar por las claves/valores de un diccionario usamos el método `items()`

```python
for clave, valor in alumno.items():
    print(clave, valor)
```

# 7. Estructuras de datos

Las estructuras de datos se pueden **combinar entre sí** para representar información más compleja.

Por ejemplo, podemos utilizar una lista de diccionarios para almacenar información sobre varios alumnos.

En este caso, cada elemento de la lista es un diccionario con los datos de un alumno.


```python
# Lista de diccionarios
alumnos = [
    {"nombre": "Ana", "nota": 8},
    {"nombre": "Luis", "nota": 4},
    {"nombre": "Marta", "nota": 6}
]

for alumno in alumnos:
    if alumno["nota"] >= 5:
        print(f"{alumno['nombre']} ha aprobado")
```

# 8. Funciones

Las **funciones** permiten agrupar un conjunto de instrucciones bajo un nombre para poder reutilizarlas.


Para definir una función utilizamos `def`.


```python
def saludar():
    print("¡Hola!")

saludar()
saludar()
```

> Una función puede recibir datos como **parámetros**.

```python
def saludar(nombre):
    print(f"¡Hola, {nombre}!")

saludar("Ana")
saludar("Luis")
```

> Una función puede devolver un **resultado** usando `return`

```python
def sumar(a, b):
    return a + b

resultado = sumar(3, 5)

print(resultado)
```

# 9. Excepciones

Durante la ejecución de un programa pueden producirse **errores**.

Por ejemplo, intentar convertir un texto que no representa un número en un `int` produce una excepción.

Podemos detectar determinados errores utilizando `try` y `except`, evitando que el programa termine inmediatamente.


```python
# Este código puede fallar?

edad = int(input("¿Cuántos años tienes? "))

print(f"Tienes {edad} años")
```

```python
# Y este?

numero = int(input("Número: "))

resultado = 10 / numero

print(resultado)
```

> Lo arreglamos usando `try` y `catch`

```python
try:
    edad = int(input("¿Cuántos años tienes? "))
    print(f"Tienes {edad} años")
except ValueError:
    print("Debes introducir un número")
```

```python
try:
    numero = int(input("Número: "))
    resultado = 10 / numero
    print(resultado)
except ZeroDivisionError:
    print("No puedes dividir entre cero")
```

```python
# Podemos comprobar más de un posible fallo

try:
    numero = int(input("Número: "))
    resultado = 10 / numero
    print(resultado)
except ValueError:
    print("Debes introducir un número")
except ZeroDivisionError:
    print("No puedes dividir entre cero")
```

# 10. Ficheros

Los programas pueden necesitar **leer o guardar información en archivos**.

Python permite abrir un fichero con `open()` y trabajar con él. Normalmente utilizaremos `with`, que se encarga de cerrar el fichero automáticamente cuando terminamos de trabajar con él.

Los modos que veremos son:

- `"r"` — leer
- `"w"` — escribir, reemplazando el contenido anterior
- `"a"` — añadir contenido al final

Guardando la información en ficheros conseguimos **almacenamiento persistente**.

```python
# Escribir en un fichero
with open("datos.txt", "w") as fichero:
    fichero.write("Hola\n")
    fichero.write("Este es mi fichero")
```

```python
# Leer un fichero
with open("datos.txt", "r") as fichero:
    contenido = fichero.read()

print(contenido)
```

```python
# Añadir otra línea
with open("datos.txt", "a") as fichero:
    fichero.write("\nOtra línea")
```

> Podemos iterar por las líneas de un fichero como sí fuera una lista:

```python
with open("datos.txt", "r") as fichero:
    for linea in fichero:
        print(linea.strip())
```

## Conexión a APIs web

```python
import requests

pokemon = input("Introduce un Pokémon: ").lower()

try:
    datos = requests.get(
        f"https://pokeapi.co/api/v2/pokemon/{pokemon}"
    ).json()

    especie = requests.get(
        f"https://pokeapi.co/api/v2/pokemon-species/{pokemon}"
    ).json()

    print(f"\n--- {datos['name'].capitalize()} ---")

    print(f"ID: {datos['id']}")
    print(f"Altura: {datos['height'] / 10} m")
    print(f"Peso: {datos['weight'] / 10} kg")

    tipos = []
    for tipo in datos["types"]:
        tipos.append(tipo["type"]["name"])

    print(f"Tipos: {', '.join(tipos)}")

    for entrada in especie["flavor_text_entries"]:
        if entrada["language"]["name"] == "es":
            texto = entrada["flavor_text"]
            print(f"\n📖 Pokédex:\n{texto}")
            break

except requests.RequestException:
    print("No se ha podido conectar con la API.")
except KeyError:
    print("No se ha encontrado ese Pokémon.")
```

## Machine learning

```python
from sklearn.datasets import load_digits
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

digits = load_digits()

modelo = KNeighborsClassifier()
modelo.fit(digits.data, digits.target)

numero = 24

plt.imshow(digits.images[numero], cmap="gray")
plt.show()

prediccion = modelo.predict([digits.data[numero]])

print(f"El modelo cree que es un {prediccion[0]}")
```
